```python
import pandas as pd
import matplotlib.pyplot as plt

students = ["Ali","Sara","Ahmed","Ayesha","Zain"]
marks = [70,90,75,85,60]
df = pd.DataFrame({"students":students,"Marks":marks})

plt.plot(df["students"], df["Marks"], marker="o")
plt.title("student marks")
plt.xlabel("student")
plt.ylabel("marks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

marks = [70, 95, 30]
performance = ["average", "high", "low"]

plt.pie(marks, labels=performance, autopct="%1.1f%%")
plt.title("Students Performance")
plt.show()
```

```python
marks = [70, 95, 30, 90, 45, 80]
performance_list = ["average", "high", "low", "high", "low", "high"]

performance_count = pd.Series(performance_list).value_counts()

plt.pie(performance_count, labels=performance_count.index, autopct="%1.1f%%")
plt.title("Students Performance")
plt.show()
```
