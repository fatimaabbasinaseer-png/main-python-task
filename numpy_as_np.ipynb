```python
import numpy as np

marks = np.array([[70,30,40,90,80],
                  [40,50,30,70,80],
                  [90,40,80,70,50]])
print(marks)
print("\nmarks + 5 =\n", marks + 5)
print("\nShape:", marks.shape)
print("Size:", marks.size)
```

```python
import numpy as np
marks = np.array([
    [70,30,40,90,80],
    [40,50,30,70,90],
    [90,40,80,70,50]])

print(marks.ndim)
print(marks.shape)
print(marks.size)
print(marks.dtype)
```

```python
import numpy as np
marks = np.array([70,30,40,90,80])
print(marks + 5)
print(marks * 5)
print(marks - 5)
print(marks / 5)
```

```python
import numpy as np
std1_marks = np.array([70,30,40,90,80])
std2_marks = np.array([40,50,30,70,90])

print(std1_marks + std2_marks)
print(std1_marks - std2_marks)
print(std1_marks * std2_marks)
print(std1_marks / std2_marks)
print(std1_marks % std2_marks)
```

```python
import numpy as np
std1_marks = np.array([70,30,40,90,80])
print(np.sqrt(std1_marks))
std2_marks = np.array([40,50,30,70,90])
print(np.power(std1_marks  ,2))
std3_marks = np.array([40,1,5,9,30,7,70,90,5])
print(np.round(std3_marks))
```
